```python
from save_load_model import load_model
import torch
from typing import Callable, List
```

```python
model, config, encode, decode, vocab, merges = load_model("models/mickiewicz_model_post.pl")
```

```python
def finish_sequence(model: torch.nn.Module ,sequence: str, new_tokens: int, encode: Callable[[str], List[int]], decode: Callable[[List[int]], str], temperature: float = 1.0):
    return decode(model.generate(torch.tensor(encode(sequence), dtype=torch.long), new_tokens=new_tokens, temperature=temperature).int().tolist())
```

```python
print(finish_sequence(model, "Niechajże się stanie wola przyrodzona", 100, encode, decode))
```
